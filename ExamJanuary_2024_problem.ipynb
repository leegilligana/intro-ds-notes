```python
import numpy as np
import matplotlib.pyplot as plt
# Part 1
def problem1_inversion(n_samples=1):
    # Distribution from part 1
    # write the code in this function to produce samples from the distribution in the assignment
    # Make sure you choose a good sampling distribution to avoid unnecessary rejections

    x = np.random.uniform(0, 1, n_samples)
    y = np.sqrt(np.log(x*(np.exp(1)-1)+1))

    # Return a numpy array of length n_samples
    return y
```

```python
import numpy as np


def problem1_inversion_2(n_samples=1):
    # Step 1: Target PDF f(x)
    def f(x):
        return 20.0 * (1.0 + 1.0 / x) * np.exp(20.0 - 1.0 / x)

    # Step 2: Chosen proposal PDF g(x) with k=20
    # g(x) = 21 * 20^21 * x^20, which is inverted by X = (1/20) * U^(1/21)
    def g(x):
        return 21.0 * (20.0**21) * (x**20)

    # Step 3: Upper bound constant M = max(f(x) / g(x))
    M = 1.0226247739749612

    # Step 4 & 5: Batch sampling and filtering
    samples = []
    batch_size = int(n_samples * 1.05) + 100  # Draw slightly more to cover ~2% rejections

    while len(samples) < n_samples:
        # Draw candidates X from g(x)
        u1 = np.random.uniform(0, 1, size=batch_size)
        candidates = (1.0 / 20.0) * (u1 ** (1.0 / 21.0))

        # Rejection threshold
        accept_probs = f(candidates) / (M * g(candidates))

        # Accept / Reject check
        u2 = np.random.uniform(0, 1, size=batch_size)
        accepted = candidates[u2 <= accept_probs]
        samples.extend(accepted)

    # Trim to exact requested count
    return np.array(samples[:n_samples])
```
